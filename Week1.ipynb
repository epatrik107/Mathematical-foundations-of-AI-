# Vector Spaces and Linear Algebra Foundations  

## Block 1 (45 min) — Theory  
**Objective:** Basic linear algebra principles and their connection to machine learning models.

### 1. Vector Spaces: $\mathbb{R}^n$
- Linear combinations, span, basis  
- Coordinates as feature representations  
- Dimension as degrees of freedom / information content  
- Linear maps ↔ matrices (models as operators)

A **vector space** $V$ over the field $\mathbb{R}$ is a set equipped with two operations:

- vector addition $+: V \times V \to V$,
- scalar multiplication $\cdot: \mathbb{R} \times V \to V$,

such that for all $u,v,w \in V$ and $\alpha,\beta \in \mathbb{R}$ the standard vector space axioms hold
(associativity, commutativity, existence of a zero vector and additive inverses, distributivity, and compatibility with scalar multiplication).

**Canonical examples:**

- $\mathbb{R}^n = \{ [x_1,\ldots,x_n]^\top \mid x_i \in \mathbb{R},\; 1 \le i \le n \}$,  
  equipped with componentwise addition and scalar multiplication.
- $\mathcal{F}(X,\mathbb{R}^m)$: the vector space of $\mathbb{R}^m$-valued functions defined on an arbitrary (abstract) set $X$,  
  with pointwise addition and scalar multiplication.
    - The space $\mathbb{R}^n$ appears as a special case by choosing $m = 1$ and $X = \{1,\ldots,n\}$.
- The vector space of real polynomials of degree at most $n$:
$$
\mathcal{P}_n
:= \left\{ p(x) = \sum_{k=0}^n a_k x^k \;\middle|\; a_k \in \mathbb{R} \right\},
$$
- The vector space of $n \times m$ real matrices, with entrywise addition and scalar multiplication:
  $$
\mathbb{R}^{n \times m}
:= \{ A = (a_{ij}) \mid a_{ij} \in \mathbb{R},\ i=1,\dots,n,\ j=1,\dots,m \},
$$


**Machine learning motivation:**  
Data points, pixel values, audio signals, feature vectors, parameters, gradients, and embeddings all live in vector spaces.  
Learning algorithms manipulate these objects using linear algebraic operations.

---

### Linear Combinations, Span, and Linear Independence

Given vectors $v_1,\dots,v_k \in \mathbb{R}^n$, a **linear combination** is any vector of the form
$$
x = \alpha_1 v_1 + \dots + \alpha_k v_k,
\quad \alpha_i \in \mathbb{R}.
$$

The **span** of $\{v_1,\dots,v_k\}$ is
$$
\mathrm{span}\{v_1,\dots,v_k\}
=
\left\{
\alpha_1 v_1 + \dots + \alpha_k v_k
\;\middle|\;
\alpha_i \in \mathbb{R}
\right\}.
$$

Vectors $v_1,\dots,v_k$ are **linearly independent** if
$$
\alpha_1 v_1 + \dots + \alpha_k v_k = 0
\quad \Longrightarrow \quad
\alpha_1 = \dots = \alpha_k = 0.
$$

Otherwise, they are **linearly dependent**.

**ML relevance:**  
Linearly dependent features introduce redundancy, numerical instability, and ill-conditioned optimization problems.

**Proposition:** For a set of vectors $\{b_1,\dots,b_d\} \subset V$, the following are equivalent:

1. Every $x \in V$ admits a unique linear combination
   $$
   x = \sum_{i=1}^d \alpha_i b_i.
   $$

2. The set $\{b_1,\dots,b_d\}$ is linearly independent and spans $V$.

3. The set is a minimal generating system of $V$.

4. The set is a maximal linearly independent subset of $V$.

A set of vectors ${b_1, \dots, b_d}$ with these properties is a **basis** of the vector space $V$. The **dimension** of a vector space is the number of vectors in any basis.

**In the ML context dimension can be:**
- number of degrees of freedom,
- information content of the representation,
- model capacity (informal),
- source of the curse of dimensionality.

Fix a basis $\{b_1,\dots,b_n\}$. Every vector $x \in V$ admits a unique coordinate representation
$$
x = \sum_{i=1}^n \alpha_i b_i
\quad \leftrightarrow \quad
(\alpha_1,\dots,\alpha_n).
$$

**Key idea:**  The vector is abstract; coordinates depend on the chosen basis.

### Linear Maps and Matrices

A mapping $T: \mathbb{R}^n \to \mathbb{R}^m$ is **linear** if
$$
T(\alpha x + \beta y) = \alpha T(x) + \beta T(y), \quad \forall x,y \in \mathbb{R}^n, \ \alpha,\beta \in \mathbb{R}.
$$

**Image (Range) and Kernel (Nullspace):**
- The **image** of $T$ is
$$
\operatorname{Im}(T) = \{ T(x) \in \mathbb{R}^m : x \in \mathbb{R}^n \}.
$$
- The **kernel** of $T$ is
$$
\ker(T) = \{ x \in \mathbb{R}^n : T(x) = 0 \}.
$$

**Dimension Theorem (Rank-Nullity):**
$$
\dim(\ker(T)) + \dim(\operatorname{Im}(T)) = n.
$$

**Matrix Representation in Finite Dimensions:**
Given bases $\{e_1,\dots,e_n\}$ of $\mathbb{R}^n$ and $\{f_1,\dots,f_m\}$ of $\mathbb{R}^m$, $T$ can be represented by a matrix $A \in \mathbb{R}^{m \times n}$ such that
$$
T(x) = A x.
$$

- The **columns of \(A\)** are the images of the basis vectors of $\mathbb{R}^n$ expressed in the $\mathbb{R}^m$ basis:
$$
A = [T(e_1) \ \cdots \ T(e_n)].
$$

- **Image of \(T\)** is spanned by the columns of \(A\):
$$
\operatorname{Im}(T) = \operatorname{span}\{ \text{columns of } A \}.
$$

- **Kernel of \(T\)** corresponds to the solution space of
$$
A x = 0.
$$

#### Eigenvalues and Eigenvectors

For a square linear map $T: \mathbb{R}^n \to \mathbb{R}^n$, a scalar $\lambda \in \mathbb{C}$ is an **eigenvalue** if there exists a nonzero vector $v$ such that
$$
T(v) = \lambda v.
$$

- $v$ is called an **eigenvector** corresponding to $\lambda$.
- The **eigenspace** of $\lambda$ is
$$
E_\lambda = \{ v \neq 0 : T(v) = \lambda v \}.
$$

A matrix $A$ is **diagonalizable** if there exists a basis of eigenvectors; then
$$
A = V \Lambda V^{-1}, \quad \Lambda = \text{diag}(\lambda_1,\dots,\lambda_n).
$$

#### Jordan Normal Form

Not all matrices are diagonalizable. Every square matrix $A \in \mathbb{C}^{n \times n}$ can be brought to **Jordan normal form** $J$:
$$
A = P J P^{-1},
$$
where $J$ is block-diagonal with **Jordan blocks**
$$
J_k(\lambda) =
\begin{bmatrix}
\lambda & 1 & 0 & \dots & 0 \\
0 & \lambda & 1 & \dots & 0 \\
\vdots & & \ddots & \ddots & \vdots \\
0 & \dots & 0 & \lambda & 1 \\
0 & \dots & \dots & 0 & \lambda
\end{bmatrix}.
$$

- Each block corresponds to an eigenvalue $\lambda$.
- Jordan form reveals the **generalized eigenspaces** and the structure of $A$ when it is not diagonalizable.

**Applications in ML and Optimization:**

- Eigenvectors are used in PCA, SVD, and spectral methods.
- Jordan decomposition is important for understanding the long-term behavior of linear dynamical systems and iterative algorithms.


### Inner Product Spaces and Orthogonality

#### General Definition of Inner Product
An **inner product** on a vector space $V$ over $\mathbb{R}$ is a function
$$
\langle \cdot, \cdot \rangle : V \times V \to \mathbb{R}
$$
satisfying for all $x, y, z \in V$ and $\alpha \in \mathbb{R}$:

1. **Linearity in the first argument**: $\langle \alpha x + y, z \rangle = \alpha \langle x, z \rangle + \langle y, z \rangle$
2. **Symmetry**: $\langle x, y \rangle = \langle y, x \rangle$
3. **Positive definiteness**: $\langle x, x \rangle \ge 0$, with equality if and only if $x = 0$.

Two vectors $x$ and $y$ are **orthogonal** if $\langle x, y \rangle = 0.$

**Cauchy–Schwarz Inequality:** For any vectors $x, y \in V$,
$$
|\langle x, y \rangle| \le \|x\| \cdot \|y\|,
$$
with equality if and only if $x$ and $y$ are linearly dependent.

#### Inner Product in Finite Dimensions: the Gram Matrix

Let $V$ be a finite-dimensional space with a fixed basis $\{b_1, \dots, b_n\}$.  
The **Gram matrix** $G \in \mathbb{R}^{n \times n}$ associated with this basis is defined by
$$
G_{ij} = \langle b_i, b_j \rangle,
$$

Then, for any vectors \(x, y \in V\) with coordinate column vectors 
$$
[x]_B = \begin{bmatrix}x_1 \\ \vdots \\ x_n\end{bmatrix}, \quad [y]_B = \begin{bmatrix}y_1 \\ \vdots \\ y_n\end{bmatrix},
$$
we have
$$
\langle x, y \rangle = [x]_B^\top G [y]_B.
$$

Here the left-hand side is the abstract inner product in $V$, while the right-hand side expresses it concretely via the coordinates in the chosen basis and the Gram matrix.  

**Properties of the Gram matrix:**

1. **Symmetric:** $G = G^\top$ 
2. **Positive semidefinite:** $v^\top G v \ge 0$ for all $v \in \mathbb{R}^n$  
3. **Positive definite** since $\{b_1, \dots, b_n\}$ is linearly independent



  
#### Orthogonal Projection as Least-Squares Solution
Given a subspace $U \subset V$ with an orthonormal basis $\{u_1, \dots, u_k\}$, the **orthogonal projection** of $v \in V$ onto $U$ is
$$
\mathrm{proj}_U(v) = \sum_{i=1}^k \langle v, u_i \rangle u_i,
$$
which also solves the least-squares problem $\min_{u \in U} \|v - u\|^2.$

#### Linear functionals

A linear functional $f$ on a real vector space $V$ is a linear map $f: V \to \mathbb{R}$.

**Riesz Representation Theorem:** For every linear functional $f$ on an inner product space $(V, \langle \cdot, \cdot \rangle)$, there exists a unique vector $w_f \in V$ such that
$$
  f(x) = \langle w_f, x \rangle \quad \forall x \in V.
$$

  
- **Identification via inner product:**  
  If $V = \mathbb{R}^n$ with an inner product $\langle x, y \rangle_M = x^\top M y$, $M \succ 0$, then for any $f \in (\mathbb{R}^n)^*$, the representing vector is
  $$
  w_f = M^{-1} \hat{f}, \quad \text{where } \hat{f} = \begin{bmatrix} f(e_1) \\ \vdots \\ f(e_n) \end{bmatrix}
  $$
  in the standard basis $\{e_i\}$.

- **ML motivation:**  
  - Linear predictors: A prediction $y = w^\top x$ is a linear functional of the input $x$.
  - For binary classification \(y \in \{0,1\}\), we pass the linear combination through a sigmoid function to obtain probabilities:
    $$
    p(y=1 \mid x) = \sigma(w^\top x + b), \quad 
    \sigma(z) = \frac{1}{1 + e^{-z}}.
    $$
  - Gradients of loss functions: The gradient $\nabla_x L(x)$ is a linear functional in the tangent space of parameters, represented as a vector in $V$ via the Riesz theorem.





### Tensor (Kronecker, Outer) Product of Matrices

Let $A \in \mathbb{R}^{m \times n}$ and $B \in \mathbb{R}^{p \times q}$. The **Kronecker product** (tensor product) $A \otimes B$ is defined as:
$$
A \otimes B =
\begin{bmatrix}
a_{11} B & a_{12} B & \dots & a_{1n} B \\
a_{21} B & a_{22} B & \dots & a_{2n} B \\
\vdots & \vdots & \ddots & \vdots \\
a_{m1} B & a_{m2} B & \dots & a_{mn} B
\end{bmatrix} \in \mathbb{R}^{mp \times nq}.
$$

Special case: the **outer product** of vectors $x \in \mathbb{R}^m$ and $y \in \mathbb{R}^n$ is:
$$
x \otimes y = xy^\top \in \mathbb{R}^{m \times n}.
$$

---

#### Basic Properties

- **Bilinearity:** $(A + C) \otimes B = A \otimes B + C \otimes B$, $\alpha A \otimes B = (\alpha A) \otimes B$
- **Mixed-product property:** $(A \otimes B)(C \otimes D) = (AC) \otimes (BD)$, if dimensions are compatible
- **Transpose:** $(A \otimes B)^\top = A^\top \otimes B^\top$

#### Vectorization

**Vectorization** is the operation of converting a matrix $X \in \mathbb{R}^{m \times n}$ into a column vector $\text{vec}(X) \in \mathbb{R}^{mn}$ by stacking its columns:
$$
\text{vec}(X) =
\begin{bmatrix}
x_1 \\ x_2 \\ \vdots \\ x_n
\end{bmatrix}, \quad x_j \text{ is the } j\text{-th column of } X
$$

**Key property linking vectorization and Kronecker product:**
$$
\text{vec}(AXB) = (B^\top \otimes A) \, \text{vec}(X)
$$

#### Why It Matters in Machine Learning

1. **Vectorization of matrix equations:**  
   Tensor products allow flattening multi-index equations into standard linear algebra form, enabling efficient gradient computation.

2. **Structured parameters:**  
   Kronecker structures reduce memory footprint in covariance matrices or large linear models, e.g., in Gaussian processes:
   $$
   \text{Cov}(X \otimes Y) = \text{Cov}(X) \otimes \text{Cov}(Y)
   $$

3. **Covariance and multilinear models:**  
   Outer and Kronecker products naturally represent interactions between multiple modes (features, time, samples) in tensor regression or probabilistic models.

### 1. Random Matrix Spectrum and Wigner Semicircle Law

**Motivation in AI:**  
In deep learning, the weight matrices of neural network layers are often initialized randomly. The spectrum (eigenvalues) of these random matrices influences the scale of activations and gradients during forward and backward passes. If the eigenvalues are too large or too small, the network can suffer from exploding or vanishing activations, which slows down or destabilizes training. 

The **Wigner semicircle law** describes the typical distribution of eigenvalues for large symmetric random matrices and helps us understand why proper weight scaling (e.g., Xavier or He initialization) is crucial.

**Wigner Semicircle Law (Theorem):**  
Let $A_n$ be an $n \times n$ symmetric random matrix with independent entries above the diagonal, each with mean 0 and variance $\sigma^2$. As $n \to \infty$, the empirical spectral distribution of $A_n / \sqrt{n}$ converges almost surely to the semicircle distribution with density

$$
\rho(x) = \frac{1}{2\pi \sigma^2} \sqrt{4\sigma^2 - x^2}, \quad |x| \le 2\sigma
$$

and $\rho(x) = 0$ for $|x| > 2\sigma$.

**Exercise:**  
Generate a large random symmetric matrix, compute its eigenvalues, and visualize the spectrum. Observe that the eigenvalue histogram approximates a semicircle as predicted by Wigner's law.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 1000  # matrix size

# Random symmetric matrix
A = np.random.randn(n, n)
A = (A + A.T) / np.sqrt(2 * n)  # normalization for semicircle law

# Eigenvalues
eigvals = np.linalg.eigvalsh(A)

# Histogram – semicircle shape
plt.hist(eigvals, bins=100, density=True, color='skyblue', alpha=0.7)
plt.title("Eigenvalue spectrum of a random symmetric matrix")
plt.xlabel("Eigenvalue")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

### 2. Gram Matrices, Feature Maps, and Kernels in AI

**Motivation in AI:**  
In machine learning, especially in kernel methods and neural networks, we often want to measure **similarities between data points** in a high-dimensional space.  
- A **feature map** $\phi(x)$ maps an input $x$ to a (possibly higher-dimensional) feature space where linear methods can capture nonlinear patterns.  
- The **Gram matrix** (or kernel matrix) contains all pairwise inner products of feature vectors:  
$$
G_{ij} = \langle \phi(x_i), \phi(x_j) \rangle
$$  
- Using **kernels** (functions that compute inner products in feature space) allows us to operate implicitly in high-dimensional spaces without computing $\phi(x)$ explicitly (the “kernel trick”).  

Common kernels:  
- **Linear kernel:** standard inner product $\langle x_i, x_j \rangle$
- **RBF / Gaussian kernel:** $\exp(-\gamma \|x_i - x_j\|^2)$, captures nonlinear similarity  

This exercise illustrates how different kernels represent data similarities and how the Gram matrix reflects the geometry of the feature space.

---

**Exercise:**  
Generate 2D random points, compute the **linear Gram matrix** and the **RBF kernel Gram matrix**, and visualize them.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import rbf_kernel

# Random 2D points
X = np.random.randn(50, 2)

# Linear Gram matrix
G_lin = X @ X.T

# RBF kernel Gram matrix
G_rbf = rbf_kernel(X, gamma=0.5)

# Visualization
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(G_lin, cmap='viridis')
plt.title("Linear Gram matrix")
plt.colorbar()

plt.subplot(1,2,2)
plt.imshow(G_rbf, cmap='viridis')
plt.title("RBF Kernel matrix")
plt.colorbar()

plt.show()
```

### 3. Word Embeddings and Gram Matrices (GloVe Example)

**Motivation in AI:**  
Word embeddings (like GloVe or Word2Vec) transform discrete words into **dense feature vectors** in a high-dimensional space.  
- Each embedding is a **feature map** $\phi(\text{word}) \in \mathbb{R}^d$ representing semantic meaning.  
- The **Gram matrix** of these embeddings, $G_{ij} = \langle \phi(x_i), \phi(x_j) \rangle$, captures **pairwise similarity** between words.  
- This is the same concept as in kernel methods: inner products in feature space quantify similarity, but here the features are learned from large text corpora.

**GloVe Feature Transformation:**  
- GloVe (Global Vectors for Word Representation) is a **word embedding model** that learns vector representations of words from **global corpus statistics**.  
- It first constructs a **co-occurrence matrix** $X$, where $X_{ij}$ counts how often word $i$ appears in the context of word $j$.  
- Each word $i$ is represented as a **dense d-dimensional vector**: $\phi(\text{word}_i) \in \mathbb{R}^d$.  
- The vectors are trained so that **inner products approximate logarithms of co-occurrence ratios**:

$$
\phi(\text{word}_i)^\top \phi(\text{word}_j) + b_i + b_j \approx \log X_{ij}
$$

- Thus, the **feature map $\phi$** maps words into a vector space where the **Gram matrix (dot product)** reflects semantic similarity.  
- After normalization, this corresponds to **cosine similarity**, which can be easily visualized.

**AI relevance:**  
- Semantic relationships like king–queen, man–woman, cat–dog are reflected in the Gram matrix.  
- The Gram matrix acts like a **kernel in feature space**, capturing similarity patterns just as linear or RBF kernels do.

2. Retrieve their pre-trained GloVe embeddings (50D).  
3. Compute the normalized Gram matrix (cosine similarity).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize
from gensim.downloader import load

# Load pre-trained GloVe embeddings (50D)
glove = load('glove-wiki-gigaword-50')

# Small word list
words = ["king", "queen", "man", "woman", "apple", "orange", "cat", "dog", "car", "bus"]
embeddings = np.array([glove[w] for w in words])

# Normalize embeddings (cosine similarity)
embeddings = normalize(embeddings)

# Gram matrix (cosine similarity)
G = embeddings @ embeddings.T

# Visualize Gram matrix
plt.figure(figsize=(8,6))
plt.imshow(G, cmap='viridis')
plt.colorbar()
plt.xticks(range(len(words)), words, rotation=45)
plt.yticks(range(len(words)), words)
plt.title("Word Embedding Similarity (Gram Matrix)")
plt.show()
```

### 4. Johnson-Lindenstrauss Lemma – Random Projection

**Motivation in AI:**  
High-dimensional data often contains many features, but many machine learning algorithms depend primarily on **pairwise distances** between points. Furthermore, the fundamental algorithms for analyzing such data often become computationally intractable as the dimensionality grows. Therefore, it is desirable to reduce the dimensionality of the data in a way that preserves its essential structure.

The **Johnson-Lindenstrauss (JL) lemma** guarantees that a **random linear projection** to a lower-dimensional space approximately preserves all pairwise distances. This enables:

- Efficient dimensionality reduction  
- Faster nearest neighbor search  
- Approximate kernel computations and feature embeddings  

---

**Johnson-Lindenstrauss Lemma (Formal Statement):**  
Let $X = \{x_1, x_2, \dots, x_m\} \subset \mathbb{R}^n$ be a set of $m$ points, and let $\varepsilon \in (0,1)$. Then there exists a mapping $f: \mathbb{R}^n \to \mathbb{R}^k$ with

$$
k = O\Big(\frac{\log m}{\varepsilon^2}\Big)
$$

such that for all $i,j \in \{1,\dots,m\}$:

$$
(1-\varepsilon)\|x_i - x_j\|^2 \le \|f(x_i) - f(x_j)\|^2 \le (1+\varepsilon)\|x_i - x_j\|^2
$$

In practice, $f$ can be taken as a **random linear projection**, e.g., $f(x) = \frac{1}{\sqrt{k}} R x$ where $R$ is a random Gaussian matrix.

---

**Exercise:**  
1. Generate a set of points in high-dimensional space.  
2. Project them to a lower-dimensional space using a **random Gaussian projection**.  
3. Compare original and projected pairwise distances.  
4. Visualize how well distances are preserved.

```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations

# Parameters
n_points = 100
high_dim = 500
low_dim = 50
np.random.seed(42)

# High-dimensional data
n_clusters = 6
centers = np.random.randn(n_clusters, high_dim) * 10
labels = np.random.choice(n_clusters, n_points)

X = centers[labels] + np.random.randn(n_points, high_dim)

# Random Gaussian matrix
R = np.random.randn(high_dim, low_dim) / np.sqrt(low_dim)
X_proj = X @ R

# Choose subset of points
idx = np.random.choice(n_points, 50, replace=False)

# Compute pairwise distances (unique pairs only)
pairs = list(combinations(idx, 2))

orig_distances = np.array([
    np.linalg.norm(X[i] - X[j]) for i, j in pairs
])

proj_distances = np.array([
    np.linalg.norm(X_proj[i] - X_proj[j]) for i, j in pairs
])

# Scatter plot: original vs projected distances
plt.figure(figsize=(6,6))
plt.scatter(orig_distances, proj_distances, alpha=0.5)
max_d = orig_distances.max()
plt.plot([0, max_d], [0, max_d], 'r--')
plt.xlabel("Original distance")
plt.ylabel("Projected distance")
plt.title("Johnson–Lindenstrauss Lemma (Orthogonal Projection)")
plt.grid(True)
plt.show()

# Relative distortion
distortion = proj_distances/ orig_distances

plt.figure(figsize=(6,4))
plt.hist(distortion, bins=40, density=True)
plt.xlabel("Distance Ratio")
plt.ylabel("Density")
plt.title("Distribution of Distance Rations")
plt.grid(True)
plt.show()
```
