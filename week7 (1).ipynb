## Complete Metric Spaces

### Motivation

Completeness is a fundamental structural property of metric spaces. Its importance lies in the fact that it allows us to decide whether a sequence converges **based solely on its internal behavior**, without any prior knowledge of its potential limit.

In many applications—analysis, probability, and numerical methods—we encounter sequences defined implicitly (e.g., by iteration). In such cases, completeness ensures that if the sequence “should” converge (in the sense that its elements get arbitrarily close to each other), then it indeed converges within the space.

---

### Cauchy Sequences

A sequence $(x_n)_{n \in \mathbb{N}}$ in a metric space $(X,d)$ is called a **Cauchy sequence** if for every $\varepsilon > 0$ there exists $N \in \mathbb{N}$ such that
$$
d(x_n, x_m) < \varepsilon \quad \text{for all } n,m \geq N.
$$

In other words, the elements of the sequence become arbitrarily close to each other as the sequence progresses.

---

### Basic Properties of Cauchy Sequences

- Every convergent sequence is Cauchy.
- Every Cauchy sequence is bounded, but not necessarily convergent.
- If a Cauchy sequence has a cluster point, then it is convergent.

---

### Complete Metric Spaces

A metric space $(X,d)$ is called **complete** if every Cauchy sequence in $X$ converges to a limit that also belongs to $X$.

---

### Completeness of $\mathbb{R}^d$

**Proposition:** The space $\mathbb{R}^d$ is complete.

#### Proof

**Case $d = 1$:**

We work in $\mathbb{R}$.

- By the **monotone subsequence lemma**, every sequence in $\mathbb{R}$ admits a monotone (non-increasing or non-decreasing) subsequence.
- A Cauchy sequence is bounded, hence any monotone subsequence is also bounded.
- Every bounded monotone sequence in $\mathbb{R}$ is convergent.
- If a Cauchy sequence has a convergent subsequence, then the whole sequence converges to the same limit.

Thus, every Cauchy sequence in $\mathbb{R}$ converges.

---

**Case $d > 1$:**

Since we are in finite dimension, all norms on $\mathbb{R}^d$ are equivalent. Therefore, it suffices to prove completeness with respect to a convenient norm. We choose
$$
\|x\|_1 = |x^1| + \cdots + |x^d|.
$$

Let $(x_n)$ be a Cauchy sequence in $\mathbb{R}^d$. Then for each coordinate $k = 1, \dots, d$,
$$
|x_n^k - x_m^k| \leq \|x_n - x_m\|_1.
$$
Thus, each coordinate sequence $(x_n^k)$ is Cauchy in $\mathbb{R}$, hence convergent by the one-dimensional case. Denote
$$
x_\infty = (x_\infty^1, \dots, x_\infty^d).
$$

Then
$$
\|x_n - x_\infty\|_1 = |x_n^1 - x_\infty^1| + \cdots + |x_n^d - x_\infty^d| \to 0 \quad \text{as } n \to \infty,
$$
since each term in this finite sum converges to zero. Therefore, $x_n \to x_\infty$, and $\mathbb{R}^d$ is complete.

### Completeness of Function Spaces

**Proposition:** The space $C(X) = C(X, \mathbb{R})$ is complete with respect to the supremum norm for any metric space $X$.

**Proof:**

Let $(f_n)_{n \in \mathbb{N}}$ be a Cauchy sequence in $C(X)$ with the supremum norm:
$$
\|f_n - f_m\|_\infty = \sup_{x \in X} |f_n(x) - f_m(x)| \to 0 \quad \text{as } n,m \to \infty.
$$

1. **Pointwise convergence:**  
   For each fixed $x \in X$, the sequence of real numbers $(f_n(x))_{n \in \mathbb{N}}$ is Cauchy in $\mathbb{R}$, since
   $$
   |f_n(x) - f_m(x)| \le \|f_n - f_m\|_\infty \to 0 \quad \text{as } n,m \to \infty.
   $$
   Hence, for each $x \in X$, the limit
   $$
   f(x) := \lim_{n \to \infty} f_n(x)
   $$
   exists in $\mathbb{R}$, defining a pointwise limit function $f: X \to \mathbb{R}$.

2. **Uniform convergence:**  
   Since $(f_n)$ is Cauchy in the supremum norm, for any $\varepsilon > 0$ there exists $N$ such that for all $n,m \ge N$,
   $$
   \|f_n - f_m\|_\infty < \varepsilon.
   $$
   Taking the limit $m \to \infty$ and using the pointwise convergence $f_m(x) \to f(x)$ gives
   $$
   \|f_n - f\|_\infty = \sup_{x \in X} |f_n(x) - f(x)| \le \varepsilon \quad \text{for all } n \ge N.
   $$
   Thus, $(f_n)$ converges **uniformly** to $f$.

3. **Continuity of the limit (sequence-based argument):**  
   Let $x \in X$ be arbitrary, and let $(x_m)_{m \in \mathbb{N}}$ be a sequence in $X$ with $x_m \to x$. Then
   $$
   |f(x_m) - f(x)| \le |f(x_m) - f_n(x_m)| + |f_n(x_m) - f_n(x)| + |f_n(x) - f(x)|.
   $$
   For any fixed $n$, the middle term $|f_n(x_m) - f_n(x)| \to 0$ as $m \to \infty$ because $f_n$ is continuous.  
   The first and last terms are bounded by $\|f_n - f\|_\infty$, which can be made arbitrarily small by choosing $n$ large enough.  
   Therefore,
   $$
   \lim_{m \to \infty} |f(x_m) - f(x)| = 0,
   $$
   proving that $f$ is continuous.

Hence, the uniform limit $f$ belongs to $C(X)$, and every Cauchy sequence in $C(X)$ converges in $C(X)$. This shows that $C(X)$ is complete.

**Remark:** As a special case, we obtain that $C([0,1])$ is complete.

### The Banach Fixed Point Theorem

**Banach fixed point theorem (contraction mapping theorem):**

Let $(X, d)$ be a complete metric space, and let $T: X \to X$ be a contraction, i.e., there exists a constant $0 \leq c < 1$ such that for all $x, y \in X$,
$$
d(T(x), T(y)) \leq c \cdot d(x, y).
$$

Then the following holds:

1. *Existence and uniqueness of the fixed point:*  
   There exists a unique point $x^* \in X$ such that $T(x^*) = x^*$.  
   That is, $x^*$ is the unique fixed point of $T$.

2. *Convergence of iterates:*  
   For any starting point $x_0 \in X$, the sequence $(x_n)$ defined by the iterative process
   $$
   x_{n+1} = T(x_n)
   $$
   converges to the fixed point $x^*$. Specifically,
   $$
   \lim_{n \to \infty} x_n = x^*.
   $$

3. *A priori error estimate:*  
   For all $n \in \mathbb{N}$, the distance between the $n$-th iterate $x_n$ and the fixed point $x^*$ satisfies
   $$
   d(x_n, x^*) \leq \frac{c^n}{1 - c} \, d(x_1, x_0).
   $$
   This gives an upper bound on the error before reaching the fixed point, depending on the initial values and the contraction constant.

4. *A posteriori error estimate:*  
   For all $n \in \mathbb{N}$, the error at the $n$-th iterate can also be estimated by
   $$
   d(x_{n+1}, x^*) \leq \frac{c}{1 - c} \, d(x_{n+1}, x_n).
   $$
   This provides an error estimate based on the difference between consecutive iterates, which can be useful during or after the iteration process.

#### Application: Gradient Descent Method

**Proposition (Convergence of Gradient Descent via Banach Fixed Point):**  

Let $f:\mathbb{R}^n \to \mathbb{R}$ be a differentiable function that is $\mu$-strongly convex, i.e., for some $\mu > 0$,
$$
\langle \nabla f(y) - \nabla f(x), y - x \rangle \ge \mu \|y - x\|^2, \quad \forall x,y \in \mathbb{R}^n.
$$
Assume furthermore that $\nabla f$ is $L$-Lipschitz:
$$
\|\nabla f(y) - \nabla f(x)\| \le L \|y - x\|, \quad \forall x,y \in \mathbb{R}^n.
$$

If the step size satisfies $\eta \in (0, 2\mu / L^2)$, then the gradient descent iteration
$$
x_{k+1} = x_k - \eta \nabla f(x_k)
$$
converges to a point $x^*$ such that $\nabla f(x^*) = 0$.

**Remarks:**
- By strong convexity of $f$, it follows that $x^*$ is the unique minimizer of $f$ on $\mathbb{R}^n$.
- Moreover, one can derive explicit upper bounds for the error $f(x_k) - f(x^*)$, but we do not pursue this here.
- These quantitative estimates will be discussed later in the optimization section.


**Proof:**

Define the mapping
$$
T(x) := x - \eta \nabla f(x).
$$

1. **Fixed point characterization:**  
A point $x^*$ is a fixed point of $T$ iff
$$
T(x^*) = x^* \quad \Longleftrightarrow \quad \nabla f(x^*) = 0.
$$

2. **Contraction property:**  
Using Lipschitz continuity and strong convexity, we estimate:
$$
\|T(x) - T(y)\|^2
= \|x - y - \eta(\nabla f(x) - \nabla f(y))\|^2.
$$

Expanding:
$$
= \|x - y\|^2 - 2\eta \langle \nabla f(x) - \nabla f(y), x - y \rangle + \eta^2 \|\nabla f(x) - \nabla f(y)\|^2.
$$

Using:
- strong monotonicity:  
  $$
  \langle \nabla f(x) - \nabla f(y), x - y \rangle \ge \mu \|x - y\|^2,
  $$
- Lipschitz gradient:  
  $$
  \|\nabla f(x) - \nabla f(y)\| \le L \|x - y\|,
  $$

we obtain:
$$
\|T(x) - T(y)\|^2
\le (1 - 2\eta \mu + \eta^2 L^2)\|x - y\|^2.
$$

Thus,
$$
\|T(x) - T(y)\| \le q \|x - y\|, \quad \text{with } q = \sqrt{1 - 2\eta \mu + \eta^2 L^2}.
$$

For $\eta \in (0, 2\mu/L^2)$, we have $q < 1$, hence $T$ is a contraction.


3. **Conclusion:**  
By the Banach fixed point theorem, $T$ has a unique fixed point $x^*$, and for any initial $x_0$,
$$
x_k \to x^*
$$
at exponential rate.

This proves convergence of gradient descent.

## Compact Metric Spaces

We have seen that for Cauchy sequences, the key step in proving convergence was to show the existence of a convergent subsequence. This motivates the notion of compactness.

Compact sets are “small” in the sense that one cannot fit infinitely many points into them without accumulation.

### Definition (Compactness)
Let $(X,d)$ be a metric space. A set $K \subseteq X$ is called **compact** if every infinite subset of $K$ has an accumulation point in $K$.

Equivalently: every sequence in $K$ admits a convergent subsequence with limit in $K$.

---

### Proposition (Compact $\Rightarrow$ complete)
Let $(X,d)$ be a metric space and $K \subseteq X$ compact. If $(x_n)_{n\in\mathbb{N}}$ is a Cauchy sequence in $K$, then it converges in $K$.

**Proof:**  
By compactness, $(x_n)$ has a convergent subsequence $x_{n_k} \to x \in K$.  
Since $(x_n)$ is Cauchy, the whole sequence converges to the same limit $x$. $\square$

---

### Remarks

- The converse is false: $\mathbb{R}$ is complete but not compact. For example, $x_n = n$ has no accumulation point.
- Every compact set is **bounded and closed** (it contains all its accumulation points).
- In finite-dimensional normed spaces, compact sets are exactly the **closed and bounded** sets.

- This fails in infinite dimensions. Example:

  Consider $C([0,1])$ with the supremum norm and the closed unit ball:
  $$
  \overline{B_1(0)} = \{f \in C([0,1]) : \|f\|_\infty \le 1\}.
  $$

  Define
  $$
  f_n(x) = \sin(2n\pi x), \quad n \in \mathbb{N}.
  $$

  Then for all $n \ne m$:
  $$
  \|f_n - f_m\|_{C([0,1])}^2 
  \ge \|f_n - f_m\|_{L^2([0,1])}^2 
  = \int_0^1 (\sin(2n\pi x) - \sin(2m\pi x))^2 dx = 1.
  $$

  Hence the sequence has no Cauchy subsequence, and therefore no convergent subsequence.

  Thus $\overline{B_1(0)}$ is not compact.

---

Compactness is therefore stronger than completeness. We need an additional strengthening of boundedness: **total boundedness**.

---

### Definition ($\varepsilon$-net)
Let $(X,d)$ be a metric space and $\varepsilon > 0$. A set $N \subseteq X$ is called an **$\varepsilon$-net** if for every $x \in X$, there exists $n \in N$ such that
$$
d(x,n) < \varepsilon.
$$

---

### Definition (Total boundedness)
A set $A \subseteq X$ is called **totally bounded** if for every $\varepsilon > 0$, there exists a **finite** $\varepsilon$-net for $A$.

### Theorem (Characterization of compactness in metric spaces)
Let $(X,d)$ be a metric space. A set $K \subseteq X$ is compact if and only if it is

- complete, and
- totally bounded.

---

## Compactness in $C(X)$

Let $(X,d)$ be a compact metric space and consider $C(X)$ with the supremum norm.

### Theorem (Arzelà–Ascoli)
A set $\mathcal{F} \subseteq C(X)$ is relatively compact (i.e. its closure is compact) if and only if

1. **uniformly bounded**:
   $$
   \sup_{f \in \mathcal{F}} \|f\|_\infty < \infty,
   $$

2. **equicontinuous**:
   $$
   \forall \varepsilon > 0 \ \exists \delta > 0 \ \text{such that} \ d(x,y)<\delta \Rightarrow |f(x)-f(y)|<\varepsilon
   \ \text{for all } f \in \mathcal{F}.
   $$

In particular, every sequence in $\mathcal{F}$ has a uniformly convergent subsequence.

## Continuous functions on compact sets

### Theorem (Weierstrass)
Let $K$ be compact and $f:K \to \mathbb{R}$ continuous. Then $f$ attains its minimum and maximum, i.e., there exist $x_{\min}, x_{\max} \in K$ such that
$$
f(x_{\min}) = \min_{x \in K} f(x), \quad
f(x_{\max}) = \max_{x \in K} f(x).
$$

---

### Definition (Uniform continuity)
A function $f:(X,d) \to (Y,\rho)$ is **uniformly continuous** if
$$
\forall \varepsilon > 0 \ \exists \delta > 0 \ \text{such that} \ d(x,y)<\delta \Rightarrow \rho(f(x),f(y))<\varepsilon,
$$
where $\delta$ depends only on $\varepsilon$, not on the points.

**Exercise:** Prove that Lipschitz functions are uniformly continuous.

---

### Theorem (Heine)
Let $K$ be a compact metric space and $f:K \to Y$ continuous. Then $f$ is **uniformly continuous**.

**Idea of proof:**  
If not, one can construct sequences $x_n, y_n$ with $d(x_n,y_n) \to 0$ but $\rho(f(x_n),f(y_n)) \not\to 0$, contradicting compactness (via subsequences) and continuity.

## Polygonal approximation via uniform continuity

Let $f:[0,1]\to\mathbb{R}$ be continuous. Since $[0,1]$ is compact, by the Heine theorem $f$ is **uniformly continuous**.

### Step 1: Use uniform continuity

Let $\varepsilon > 0$. By uniform continuity, there exists $\delta > 0$ such that
$$
|x-y| < \delta \;\Rightarrow\; |f(x)-f(y)| < \varepsilon.
$$

Choose a partition
$$
0 = x_0 < x_1 < \dots < x_N = 1
$$
such that $x_{i+1} - x_i < \delta$ for all $i$.

---

### Step 2: Define the polygonal (piecewise linear) approximation

Define $g:[0,1]\to\mathbb{R}$ by linear interpolation:
for $x \in [x_i, x_{i+1}]$,
$$
g(x) = f(x_i) + \frac{f(x_{i+1}) - f(x_i)}{x_{i+1} - x_i}(x - x_i).
$$

---

### Step 3: Uniform approximation

For any $x \in [x_i, x_{i+1}]$,
$$
|f(x) - f(x_i)| < \varepsilon, \quad |f(x) - f(x_{i+1})| < \varepsilon.
$$

Since $g(x)$ is a convex combination of $f(x_i)$ and $f(x_{i+1})$, it follows that
$$
|f(x) - g(x)| < \varepsilon.
$$

Hence
$$
\|f - g\|_\infty < \varepsilon.
$$

**Conclusion:** every continuous function on $[0,1]$ can be uniformly approximated by polygonal (piecewise linear) functions.

---

## Polygonal functions as ReLU neural networks

Let $g:[0,1]\to\mathbb{R}$ be a piecewise linear function with breakpoints
$$
0 = x_0 < x_1 < \dots < x_N = 1.
$$

Define slopes on each interval:
$$
m_i = \frac{g(x_{i+1}) - g(x_i)}{x_{i+1} - x_i}.
$$

Then $g$ can be written in the form
$$
g(x) = a_0 + a_1 x + \sum_{i=1}^{N-1} c_i (x - x_i)_+,
$$
where
$$
(x - x_i)_+ = \max\{0, x - x_i\}
$$
is the ReLU function, and
$$
c_i = m_i - m_{i-1}.
$$

---

### Interpretation as a neural network

Let $\sigma(x) = \max\{0,x\}$ be the ReLU activation.

Then
$$
g(x) = a_0 + a_1 x + \sum_{i=1}^{N-1} c_i \, \sigma(x - x_i).
$$

This is exactly a **single hidden layer neural network**:

- input: $x$
- hidden neurons: $\sigma(x - x_i)$
- output: linear combination

---

### Final conclusion

- Continuous functions on $[0,1]$ can be uniformly approximated by polygonal functions.
- Polygonal functions can be represented exactly by one-layer ReLU neural networks.

Therefore, **one-hidden-layer ReLU networks are dense in $C([0,1])$**.

## Approximation with polynomial functions


### Step 1: Approximation of $\sqrt{x}$ on $[0,1]$

Consider the following recursively defined sequence of functions on $[0,1]$:
$$
p_0(x) = 0, \quad p_{n+1}(x) = p_n(x) + \frac{1}{2}\bigl(x - p_n^2(x)\bigr), \quad n \in \mathbb{N}.
$$

It can be shown by induction that for all $n \in \mathbb{N}$,
$$
0 \le \sqrt{x} - p_n(x) \le \frac{2\sqrt{x}}{2 + n\sqrt{x}}, \quad x \in [0,1],
$$
which immediately implies that
$$
\|p_n - \sqrt{x}\|_{C([0,1])} \le \frac{2}{n} \to 0 \quad \text{as } n \to \infty.
$$

---

### Step 2: Approximation of $|x|$

Observe that
$$
|x| = \sqrt{x^2}.
$$

Since $x \mapsto x^2$ maps $[-1,1]$ into $[0,1]$, we can approximate $\sqrt{y}$ uniformly on $[0,1]$ by polynomials $p_n(y)$, and define
$$
q_n(x) := p_n(x^2).
$$

Then each $q_n$ is a polynomial, and
$$
\sup_{x \in [-1,1]} |q_n(x) - |x|| 
= \sup_{x \in [-1,1]} |p_n(x^2) - \sqrt{x^2}|
\le \sup_{y \in [0,1]} |p_n(y) - \sqrt{y}| \to 0.
$$

Hence $|x|$ can be uniformly approximated by polynomials on $[-1,1]$ (and in particular on $[0,1]$).

---

### Step 3: Approximation of ReLU

Recall
$$
\mathrm{ReLU}(x) = \frac{x + |x|}{2}.
$$

Let $q_n$ be polynomial approximations of $|x|$. Then define
$$
r_n(x) := \frac{x + q_n(x)}{2}.
$$

Each $r_n$ is a polynomial, and
$$
\|r_n - \mathrm{ReLU}\|_\infty 
\le \frac{1}{2} \|q_n - |x|\|_\infty \to 0.
$$

Thus $\mathrm{ReLU}$ is uniformly approximable by polynomials on compact intervals.

---

## Conclusion: Density of polynomials in $C([0,1])$

From the previous result:

- Every $f \in C([0,1])$ can be uniformly approximated by polygonal (piecewise linear) functions.
- Every polygonal function can be written using finitely many ReLU functions:
  $$
  g(x) = a_0 + a_1 x + \sum_{i=1}^{N-1} c_i \, \mathrm{ReLU}(x - x_i).
  $$

- Each $\mathrm{ReLU}$ can be uniformly approximated by polynomials.

Therefore, finite linear combinations of ReLU functions can be uniformly approximated by polynomials, hence:

$$
\overline{\mathbb{P}}^{\|\cdot\|_\infty} = C([0,1]).
$$

That is, **every continuous function on $[0,1]$ can be uniformly approximated by polynomials**.
